```python
from langchain_ollama import ChatOllama
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser #Parser --> TypeCasting
import os
```

```python
gemini_model = ChatGoogleGenerativeAI(
    model = 'gemini-3.8-flash',
    api_key = os.environ['Gemini_api'],
    temperature = 0
)

qwen_model = ChatOllama(
    model = 'qwen-2.5:1.5b',
    temperature = 0
)
```

## **LCEL --> Lang Chain Expression Language**

```python
# Step 1: Creating Prompt Temlplate,
prompt_template = ChatPromptTemplate.from_messages([
    ('system',"You're a helpful assitant, Detect the native communication Language from the context return only the detected language"),
    ('human',"{prompt}")
]) # ChatPromptTemplate has 3 roles --> human , ai , system
```

```python
# Step 2 : Creating LLM Model,
gemini_model = ChatGoogleGenerativeAI(
    model = 'gemini-3.8-flash',
    api_key = os.environ['Gemini_api'],
    temperature = 0
)

qwen_model = ChatOllama(
    model = 'qwen2.5:1.5b',
    temperature = 0
)
```

```python
# Step 3 : Create Parser,
parser = StrOutputParser() 
# StrOutputParser(), DictOutputParser()
```

```python
language_chain =  prompt_template | gemini_model | parser 
response = language_chain.invoke({'prompt' : 'Machine learning la neraya algorithms iruku,adhula Logistic Regression Sentence Based Classification ku payan padum'})
```
