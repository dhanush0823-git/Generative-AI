```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser
from langchain_ollama import ChatOllama
import os
```

## **Combining Multiple Prompt Template**

```python
# Step 1:
prompt_template = ChatPromptTemplate([
    ('system',
     "You're a helpful assitant."
     "Detect the language from the given text and translate to english language"
     "Return Only Detected Language andd Translated Text"),
     ("human","{text}")
])
```

```python
# Step 2:
gemini_model = ChatGoogleGenerativeAI(
    model = 'gemini-3.7-flash',
    api_key = os.environ['Gemini_api'],
    temperature = 0
)

qwen_model = ChatOllama(
    model = 'qwen2.5:1.5b',
    temperature=0
)
```

```python
# Step 3:
parser = StrOutputParser()
```

```python
language_chain1 = prompt_template | gemini_model | parser
```

```python
prompt_template2 = ChatPromptTemplate([
    ('system',
     "You're a helpful assitant"
     "Summarize the given text and Detected language from the translated text and return only the Detected language and the translated text"
     ),
    ('human','{text}')
])
```

```python
language_chain2 = prompt_template2 | gemini_model | parser
```

```python
full_chain = language_chain1 | language_chain2
```

```python
response = full_chain.invoke({'text' : 'Machine learning la neraya algorithms iruku,adhula Logistic Regression Sentence Based Classification ku payan padum' })
```
