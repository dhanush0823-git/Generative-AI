```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_ollama import ChatOllama
import os
```

## ***Combining Multiple Prompt Template (LCEL)***

## **Data Extraction**

```python
data = PyPDFLoader('S_Dhanush.pdf').load()
```

```python
data1 = ""
for chunk in data:
    data1 += chunk.page_content
data = data1
```

```python
# Do we have to pass the same data in all the prompt template
```

### **Prompt Template1**

```python
# Step 1:
prompt_template = ChatPromptTemplate([
    ('system',"You're a Helpful Assitant,Check the ATS score of the Resume Content based on Data Scientist Role"),
    ('human','{dataa}') 
])
```

### **Model Creation**

```python
## Step 2:
gemini_model = ChatGoogleGenerativeAI(
    model = 'gemini-3.7-flash',
    api_key = os.environ['Gemini_api'],
    temperature = 0
)

qwen_model = ChatOllama(
    model = 'qwen2.5:1.5b',
    temperature=0
)
```

### **Creating Parser**

```python
# Step 3:
parser = StrOutputParser()
```

```python
language_chain1 = prompt_template | qwen_model | parser
```

```python
prompt_template2 = ChatPromptTemplate([
    ('system',"You're a Helpful Assistant"
     "Give me the Suggestion for my resume to improve my ATS Score"
     "Return the ATS Score and the Structured Suggestion"),
     ('human','{dataa}')
])
```

```python
language_chain2 = prompt_template2 | qwen_model | parser
```

## **LCEL --> LangChain Expression Language**

### * **Combining two prompt template**

```python
full_chain = language_chain1 | language_chain2 # Works Squentially 
```

```python
response = full_chain.invoke({'dataa' : data})
```
