```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_ollama import ChatOllama
import os
```

```python
gemini_model = ChatGoogleGenerativeAI(
    model = 'gemini-3.5-flash',
    api_key = os.environ['Gemini_api'],
    temperature = 0
)
```

```python
prompt_template = ChatPromptTemplate.from_messages([
    ("system","You're an Helpful Assitant."),
    MessagesPlaceholder(variable_name="history"), # it creates a location
    ("human","{input}")
])
```

```python
chain = prompt_template | gemini_model | StrOutputParser()
```

```python
memory = {}

def get_session_history(session_id):
    if session_id not in memory:
        memory[session_id] = InMemoryChatMessageHistory()
    return memory[session_id]

chain_with_memory = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key='input',
    history_messages_key='history'
)

user1 = chain_with_memory.invoke(
    {"input" : "My name is Alice."},
    config = {"configurable":{"session_id":"user_1"}}
)

user2 = chain_with_memory.invoke(
    {"input" : "What is My Name? "},
    config = {"configurable":{"session_id":"user_1"}}
)
```

```python
while True:
    user_input = input("Enter Your Query ?")
    if user_input.lower() == "exit":
        break
    response = chain_with_memory.invoke(
    {"input" : user_input},
    config = {"configurable":{"session_id":"user_1"}}
    )
    print(f"AI Response : {response}")
```
