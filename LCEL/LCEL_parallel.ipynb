```python
# To run LCEL parallely we use RunnableParallely
# web searching tools(DuckDuckGo) - research 
# ChatGPT --> Web Searching
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.runnables import RunnableParallel
from langchain_ollama import ChatOllama
import os
```

```python
data = PyPDFLoader('S_Dhanush.pdf').load()
```

```python
data1 = ""
for chunk in data:
    data1 += chunk.page_content
data = data1
```

```python
gemini_model = ChatGoogleGenerativeAI(
    model = 'gemini-3.7-flash',
    api_key = os.environ['Gemini_api'],
    temperature = 0
)

qwen_model = ChatOllama(
    model = 'qwen2.5:1.5b',
    temperature=0
)
```

```python
parser = StrOutputParser()
```

## **LCEL 1 --> Extracting the Skills from Resume**

```python
skill_prompt = ChatPromptTemplate.from_messages([
    ("system","Extract the skills from the resume"),
    ('human',"resume : {resume}")
])

skill_chain = skill_prompt | gemini_model | parser
```

## **LCEL 2 --> Summarize the about project/exp from the resume**

```python
summarize_prompt = ChatPromptTemplate.from_messages([
    ('system','Summarize the Experience from the resume'),
    ('human','resume : {resume}')
])

summarize_chain = summarize_prompt | gemini_model | parser
```

## **LCEL 3 --> Find the Missing Skills from the resume**

```python
missing_skills = ChatPromptTemplate.from_messages([
    ('system','For Data Science Role find the missing skills from the resume.'),
    ('human',"resume : {resume}")
])
missing_chain = missing_skills | gemini_model | parser
```

## **RunnableParallel --> To run the LCEL parallely**

```python
final_chain = RunnableParallel(
    skills = skill_chain,
    summarize = summarize_chain,
    missing_skills = missing_chain
)
```

```python
sample_resume = data
```

```python
result = final_chain.invoke({'resume' : sample_resume})
```

```python
print(result['skills']) # skills,summarize,missing_skills
```
