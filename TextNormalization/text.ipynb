```python
import re
import string
import contractions
from textblob import TextBlob
import emoji
import nltk
from nltk.stem import PorterStemmer,LancasterStemmer,SnowballStemmer,WordNetLemmatizer
import spacy

class TextNormalization:

    def __init__(self,data):
        self.data = data
        self.start()

    def start(self):
        print(
            "1.Removing Spaces\n2.Removing Special Character\n3.Removing Punctuations\n4.Correcting the words\n5.Expanding the words\n6.Convert to Lowercase"
        )
        choice = input("Do you want to clean data?").lower()
        if choice == "yes":
            user = input("Do you wanna use NLP techniques ? Yes/No").lower()
            if user == "no":
                self.text_normalize()
                print("Yeah! I have performed Text Normalization!\n")
            elif user == "yes":
                user_input = input("Stemming / Lemmatization ?").lower()
                if user_input == "stemming":
                    print("1.Porter \n2.Snowball \n3.Lancaster ")
                    self.stem_choice = input("Which Stemming do you want to perform in the options ?").lower()
                    if self.stem_choice == "porter":
                        self.nltk_porter()
                    elif self.stem_choice == "snowball":
                        self.nltk_snowball()
                    elif self.stem_choice == "lancaster":
                        self.nltk_lancaster()
                    else:
                        print("Enter a Valid Technique !")
                    
                elif user_input == "lemmatization":
                    self.lemma_choice = input("Using nltk / spacy").lower()
                    if self.lemma_choice == "nltk":
                        self.nltk_lemmatize()
                    elif self.lemma_choice == 'spacy':
                        self.spacy_lemmatize()
                    else:
                        print("Enter a valid technique !")

                else:
                    print("Enter a valid NLP !")
        else:
            return False

    def text_normalize(self):
        self.converting_lower()
        self.expanding_words()
        self.correcting_the_words()
        self.removing_spaces()
        self.removing_punctuations()
        self.removing_emojis()
        
    def converting_lower(self):
        self.data = self.data.lower()

    def removing_spaces(self):
        self.data = re.sub(r'\s{2,}','',self.data)
    
    def removing_punctuations(self):
        var = string.punctuation
        self.data = re.sub(f'[{var}]',' ',self.data)

    def correcting_the_words(self):
        self.data = contractions.fix(self.data)

    def expanding_words(self):
        Obj_Textblob = TextBlob(self.data).correct()
        self.data = str(Obj_Textblob)
    
    def removing_emojis(self):
        self.data = emoji.replace_emoji(self.data,'>')

    def nltk_porter(self):
        print("As per your input we are performing --> ",self.stem_choice)
        temp = self.data.split()
        Porter = PorterStemmer()
        for word in temp:
            print(word," --> ",Porter.stem(word))

    def nltk_snowball(self):
        print("As per your input we are performing --> ",self.stem_choice)
        temp = self.data.split()
        Snowball = SnowballStemmer('english')
        for word in temp:
            print(word,"-->",Snowball.stem(word))

    def nltk_lancaster(self):
        print("As per your input we are performing --> ",self.stem_choice)
        temp = self.data.split()
        Lancaster = LancasterStemmer()
        for word in temp:
            print(word,"-->",Lancaster.stem(word))

    def nltk_lemmatize(self):
        def part_of_speech(word):
            if word[0] == 'V':
                return 'v'
            elif word[0] =='N':
                return 'n'
            elif word[0]=='A':
                return 'a'
            else:
                return 'n'
        print("As per your input we are performing --> ",self.lemma_choice)
        temp = self.data.split()
        wordnet = WordNetLemmatizer()
        token_with_pos = nltk.pos_tag(temp)
        for token,pos in token_with_pos:
            print(token,"-->",wordnet.lemmatize(token,part_of_speech(pos)))

    def spacy_lemmatize(self):
        print("As per your input we are performing --> ",self.lemma_choice)
        nlp = spacy.load('en_core_web_sm')
        data = self.data.lower()
        tokens = nlp(data)
        for token in tokens:
            print(token,"-->",token.lemma_)

data = "It is a foundation model which is already trained with vast amount of data to perform text generation tasks."
obj  = TextNormalization(data)
print(obj.data)

```
