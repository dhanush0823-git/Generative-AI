### What is Lemmatization ?

* It is a NLP technique which reduces the words to its root form as by dictionary
* It is not following any set of rules 

```python
from nltk.stem  import WordNetLemmatizer
```

```python
tokens = ['running', 'runs', 'ran', 'studies', 'studying', 'studied', 'better', 'best', 'mice', 'geese', 'children', 'cars', 'leaves', 'went', 'was', 'has', 'eating', 'ate', 'eaten', 'playing', 'played', 'cities', 'wolves', 'feet', 'teeth', 'quickly', 'happily', 'boxes', 'writing', 'wrote']
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
import nltk
```

```python
# nltk.download('wordnet') --> wordnet for lemmatization
```

```python
for word in tokens:
    print(word,"-->",lemmatizer.lemmatize(word))   #pos--> Part of speech
```

```python
token_with_pos = nltk.pos_tag(tokens) #lemmatization works with Part of Speech
```

```python
'''If the pos starts with,
n = Noun
v = Verb
a = adverb
a = adjective
s = satellite 
[(word,pos)] --> pos - Part Of Speech
''' 
token_with_pos 
```

```python
def part_of_speech(word):
    # word = token_with_pos[0][1]
    if word[0] == 'V':
        return 'v'
    elif word[0] =='N':
        return 'n'
    elif word[0]=='A':
        return 'a'
    else:
        return 'n'
print(part_of_speech('VVA'))
```

```python
for token,pos in token_with_pos:
    print(token,"-->",lemmatizer.lemmatize(token,part_of_speech(pos)))
```
