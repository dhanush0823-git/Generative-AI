### What is Stemming ?

* It is a NLP technique which reduces the words to its root form by removing prefix and suffix characters.
* it removes Prefix and Suffix Characters based on some rules or instructions given to the stemming modules..

```python
# Lancaster stemmer --> Snowball stemmer
```

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
```

```python
words = [
    "running", "runs", "runner", "studies", "studying",
    "studied", "happily", "happiness", "relational", "connections",
    "connected", "connecting", "organizations", "organization", "organized",
    "organizing", "playing", "played", "plays", "beautifully",
    "beauty", "cats", "ponies", "caresses", "flies",
    "dying", "lying", "agreed", "disabled", "matting",
    "mating", "meeting", "milling", "messing", "generously",
    "generosity", "traditional", "nationalization", "adjustments", "replacement",
    "quickly", "faster", "slowest", "friendships", "friendliness",
    "hopeful", "hopeless", "reconsidered", "universal", "automation"
]
```

```python
for word in words:
    print(f" {word} --> {stemmer.stem(word)}")
```

### **Snowball Stemmer**

```python
from nltk.stem import SnowballStemmer
```

```python
snowball = SnowballStemmer('english') #ignore_stopwords=True --> no need we are removing before this process
```

```python
words = [
    "running", "runs", "runner", "studies", "studying",
    "studied", "happily", "happiness", "relational", "connections",
    "connected", "connecting", "organizations", "organization", "organized",
    "organizing", "playing", "played", "plays", "beautifully",
    "beauty", "cats", "ponies", "caresses", "flies",
    "dying", "lying", "agreed", "disabled", "matting",
    "mating", "meeting", "milling", "messing", "generously",
    "generosity", "traditional", "nationalization", "adjustments", "replacement",
    "quickly", "faster", "slowest", "friendships", "friendliness",
    "hopeful", "hopeless", "reconsidered", "universal", "automation"
] 
```

```python
# Difference between PorterStemmer -- SnowballStemmer (only english  -- different languages)
for word in words:
    print(f" {word} --> {snowball.stem(word)}")
```

```python
from nltk.stem import LancasterStemmer
```

```python
lancaster = LancasterStemmer()
```

```python
for word in words:
    print(f" {word} --> {lancaster.stem(word)}")
```
