```python
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
data = '''What is RAG?- It stands for Retrieval Augmented Generation,it is an Ai technique which retieives the relevant data from the external sources 
and LLM generates Structured output- RAG we use for custom chatbots Example : Customer Support Chatbots
'''.lower()
tokens = nlp(data)
```

```python
for token in tokens:
    print(token,"-->>",token.lemma_) # by default it uses POS--> Part of speech
```
