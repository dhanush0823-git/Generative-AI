```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_ollama import ChatOllama
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_community.tools import DuckDuckGoSearchRun
from langchain.agents import create_agent
from langchain.tools import tool
import os
```

```python
@tool
def movie_suggestion(text : str) -> str:
    "Check Which actor name the user enter act according to the condition greet them with related message in actor's style"
    text = text.lower()
    if text in ["str",'simbu']:
        str_movies(text)
    elif text in ['vijay','thalapathy']:
        vijay_movies(text)
    elif text in ['Ajith','Thala']:
        ajith_movies(text)
    elif text in ['Thalaivar','Superstar',"Rajni"]:
        rajni_movies(text)
    else:
        pass #search engine
@tool
def str_movies(text : str) -> str:
    "Check the User Enters Str Movies Suggestions."
    movie = ["Thotti Jaya","Vallavan","Manmadhan","Silambattam","Vaanam"]
    return movie

@tool
def vijay_movies(text : str) -> str:
    "Check the User Enters Vijay Movies Suggestions."
    movie = ["Sachien","Thirumala","Sivakasi","Thirupachi","Rasigan"]  
    return movie
@tool
def ajith_movies(text : str) -> str:
    "Check the User Enters Ajith Movies Suggestions."
    movie = ["Citizen","Red","Mankatha","Aarambam","Amarkalam","Vaali"]  
    return movie
@tool
def rajni_movies(text : str) -> str:
    "Check the User Enters Rajini Movies Suggestions."
    movie = ["Thalapathi","Padayappa","Basha","Kabali","Kaala"]  
    return movie

tools = [movie_suggestion,str_movies,vijay_movies,ajith_movies,rajni_movies]

gemini_model = ChatGoogleGenerativeAI(
    model = 'gemini-3.6-flash',
    api_key = os.environ['Gemini_api'],
    temperature = 0
)

agent = create_agent(gemini_model, tools)

response = agent.invoke({
    "messages" : [
        "user","I need Ajith Movie Suggestion."
    ]
})
```
