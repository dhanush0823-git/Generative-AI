```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
from langchain.agents import create_agent 
import os 
```

```python
@tool #tool(get_word_count)
def get_word_count(text : str) -> str :
    """ Count the number of words in the given text.""" #DocString --> Important
    return f"{len(text.split())} words!"

@tool
def reverse_text(text : str) -> str :
    """Reverse the given text."""
    return text[::-1]

tools = [reverse_text,get_word_count]

#---------------LLM-------------------
gemini_model = ChatGoogleGenerativeAI(
    model = 'gemini-3.6-flash',
    api_key = os.environ['Gemini_api'],
    temperature = 0
)

#--------------Create Agents-------------
agent = create_agent(gemini_model, tools) # 
'''
Here gemini_model --> Navingates the prompt to choose the string from the prompt.
* It should be the first Argument 
Because,
It Navigates the agent based on the prompt to choose which tool for the prompt
'''
#--------------Run Agent-----------------
response = agent.invoke({
    "messages" : [
        "user","How many words are in 'Hello World this is LangChain'?"
    ]
})
```

```python
print(response['messages'][-1].content[0]['text'])
```
