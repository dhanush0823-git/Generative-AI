```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
from langchain.agents import create_agent 
from langchain_community.tools import DuckDuckGoSearchRun
from langchain_ollama import ChatOllama
import os 
```

```python
# pip install duckduckgo-search --> Search Enige
```

```python
# pip install -U ddgs
```

```python
search = DuckDuckGoSearchRun()

tools = [
    search
]
# --------------LLM Model-----------------
gemini_model = ChatGoogleGenerativeAI(
    model = 'gemini-3.5-flash',
    api_key = os.environ['Gemini_api'],
    temperature = 0
)
qwen_model = ChatOllama(
    model = 'qwen2.5:1.5b',
    temperature=0
)
# --------------Create Agent---------------
agent = create_agent(qwen_model, tools)

# response = agent.invoke({
#     "messages" : [
#         "user","Tell me Today's Gold price use this website 'https://www.goodreturns.in/gold-rates/bangalore.html'"
#     ]
# })
```
