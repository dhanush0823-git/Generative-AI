```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
from langchain.agents import create_agent 
import os 
```

```python
@tool
def prime_number(num : int) -> int:
    "Check the given number is prime number or not."
    cnt = 0
    for i in range(2,num//2):
        if num%i == 0:
            cnt+=1

    if cnt == 0:
        return "Prime Number"
    else:
        return "Not a Prime Number"
@tool
def perfect_number(num : int) -> int:
    "Check the given Number is Perfect Number or not"
    sum_ = 0
    for i in range(1,num//2):
        if num%i == 0:
            sum_ += i
    if sum_ == num:
        return "Perfect Number"
    else:
        return "Not a perfect Number"

tools = [prime_number,perfect_number]

gemini_model = ChatGoogleGenerativeAI(
    model = 'gemini-3.6-flash',
    api_key = os.environ['Gemini_api'],
    temperature = 0
)

agent = create_agent(gemini_model, tools)

response = agent.invoke({
    "messages" : [
        "user","Number 20 is a perfect number or not?"
    ]
})
```

```python
print(response['messages'][-1].content[0]['text'])
```
