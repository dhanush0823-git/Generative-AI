### Converting char to lower case

```python
import re
import string
```

```python
var = string.punctuation
```

```python
data = open('ragflow.txt','r',encoding='utf-8').read()
```

```python
var
```

```python
reg = re.sub(f'[{var}]',' ',data)
```

```python
print(reg)
```

```python
reg1 = re.sub(r'\s{2,}','',data)
```

```python
print(reg1)
```

```python
# Removing and replacing the emojis with ?
data = open('ragflow.txt',encoding='utf-8').read() # encoding='utf-8' --> for getting the emojis
print(data)
```

```python
pattern = r'[\U0001F000-\U0001FAFF\u2600-\u27BF\u2300-\u23FF\u2B00-\u2BFF\uFE0F\u200D\u20E3]+'

reg3 = re.sub(pattern, '?', data)

print(reg3)
```
